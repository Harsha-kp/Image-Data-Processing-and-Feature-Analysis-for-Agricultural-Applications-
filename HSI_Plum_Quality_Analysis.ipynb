```
import os
import zipfile
import gc
import joblib
import numpy as np
import scipy.io as sio
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
import json

from scipy.signal import savgol_filter
from skimage.filters import threshold_otsu
from skimage.morphology import remove_small_objects, binary_opening, binary_closing, disk
from skimage.measure import label, regionprops

from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import (
    classification_report, confusion_matrix, accuracy_score,
    precision_score, recall_score, f1_score, roc_auc_score,
    matthews_corrcoef, cohen_kappa_score, roc_curve, auc
)

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, models, regularizers
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau, CSVLogger
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.initializers import HeNormal

# Enable mixed precision for faster training
# tf.keras.mixed_precision.set_global_policy('mixed_float16') # Disabled as per user request to fix MaxPool3D kernel error

# Set seeds for reproducibility
SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)

from google.colab import drive
drive.mount('/content/drive')


# ============================================================
# 1. CONFIGURATION
# ============================================================

class Config:
    """Centralized configuration for reproducibility and easy tuning"""

    # Paths
    ZIP_PATH = "/content/drive/MyDrive/Dataset-Plum 2 (1).zip"
    EXTRACT_PATH = "/content/Dataset-Plum-Improved"
    OUTPUT_DIR = "/content/drive/MyDrive/HSI_Results"

    # Dataset
    CLASS_NAMES = ["PlumBefore", "Plum12hr", "Plum24hr", "Plum48hr"]

    # Preprocessing
    ROI_MIN_SIZE = 150
    SG_WINDOW = 11
    SG_POLYORDER = 2
    PCA_COMPONENTS = 50
    PCA_VARIANCE_THRESHOLD = 0.95

    # Patch extraction (multi-scale)
    PATCH_SIZES = [21, 25, 29]  # Multi-scale patches
    PATCHES_PER_IMAGE = 15
    USE_MULTI_SCALE = True

    # Model architecture
    CNN_FILTERS = [16, 32, 48]  # Lightweight filters
    DROPOUT_RATE = 0.35
    L2_REG = 0.001
    SE_RATIO = 8

    # Training
    BATCH_SIZE = 16
    EPOCHS = 10
    LEARNING_RATE = 0.0005
    LR_DECAY = 0.0001
    EARLY_STOP_PATIENCE = 20
    VAL_SPLIT = 0.2

    # Hyperparameter tuning
    USE_TUNER = False
    TUNER_EPOCHS = 30
    TUNER_TRIALS = 10

    # Augmentation
    USE_AUGMENTATION = True
    AUGMENT_ROTATION = True
    AUGMENT_FLIP = True

    # Evaluation
    USE_CROSS_VALIDATION = False
    CV_FOLDS = 5

    # Misc
    RANDOM_STATE = 42
    MAX_PIXELS_PER_IMAGE_PCA = 600

    def __init__(self):
        os.makedirs(self.OUTPUT_DIR, exist_ok=True)
        os.makedirs(self.EXTRACT_PATH, exist_ok=True)


cfg = Config()


# ============================================================
# 2. DATA LOADING
# ============================================================

def extract_dataset(zip_path, extract_path):
    """Extract dataset from ZIP"""
    if os.path.exists(extract_path) and len(os.listdir(extract_path)) > 0:
        print("Dataset already extracted.")
        return extract_path

    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_path)

    print("ZIP extracted.")
    return extract_path


def find_dataset_root(extract_path, class_names):
    """Find root directory containing class folders"""
    for root, dirs, files in os.walk(extract_path):
        if all(cls in dirs for cls in class_names):
            return root
    raise ValueError("Class folders not found.")


def collect_file_paths(dataset_path, class_names):
    """Collect all .mat file paths and labels"""
    file_paths = []
    labels = []

    for cls in class_names:
        folder = os.path.join(dataset_path, cls)
        files = sorted([f for f in os.listdir(folder) if f.endswith(".mat")])

        for f in files:
            file_paths.append(os.path.join(folder, f))
            labels.append(cls)

    return np.array(file_paths), np.array(labels)


# ============================================================
# 3. HSI CUBE LOADING
# ============================================================

def load_hsi_cube(mat_path):
    """Load HSI cube from MAT file with multiple key fallbacks"""
    mat = sio.loadmat(mat_path)

    # Priority keys
    possible_keys = ["HSImg_cut", "cube", "data", "HSI", "img", "HSI_cut", "hyperspectral"]

    for key in possible_keys:
        if key in mat and len(mat[key].shape) == 3:
            return mat[key].astype(np.float32)

    # Fallback: find first 3D array
    for key in mat.keys():
        if not key.startswith("__"):
            arr = mat[key]
            if len(arr.shape) == 3:
                return arr.astype(np.float32)

    raise ValueError(f"No 3D cube found in {mat_path}")


def minmax_normalize(cube):
    """Min-max normalization per band"""
    cube = cube.astype(np.float32)
    cube_min = np.min(cube, axis=(0, 1), keepdims=True)
    cube_max = np.max(cube, axis=(0, 1), keepdims=True)
    return (cube - cube_min) / (cube_max - cube_min + 1e-8)


# ============================================================
# 4. ENHANCED PREPROCESSING
# ============================================================

def get_roi_mask(cube, min_size=None):
    """Improved ROI extraction with adaptive thresholding"""
    if min_size is None:
        min_size = cfg.ROI_MIN_SIZE

    h, w, b = cube.shape
    band_index = min(100, b - 1)
    band = cube[:, :, band_index]

    # Adaptive thresholding
    try:
        thresh = threshold_otsu(band)
        mask = band > thresh
    except:
        mask = band > np.percentile(band, 60)

    # Morphological operations
    mask = remove_small_objects(mask.astype(bool), min_size=min_size)
    mask = binary_opening(mask, disk(2))
    mask = binary_closing(mask, disk(3))

    # Get largest connected component
    labeled = label(mask)
    regions = regionprops(labeled)

    if len(regions) == 0:
        return np.ones((h, w), dtype=bool)

    largest_region = max(regions, key=lambda r: r.area)
    final_mask = labeled == largest_region.label

    return final_mask


def sg_smoothing(cube, window_length=None, polyorder=None):
    """Savitzky-Golay smoothing with adaptive window"""
    if window_length is None:
        window_length = cfg.SG_WINDOW
    if polyorder is None:
        polyorder = cfg.SG_POLYORDER

    h, w, b = cube.shape
    if b < window_length:
        window_length = b if b % 2 == 1 else b - 1
    if window_length < 3:
        window_length = 3

    return savgol_filter(
        cube,
        window_length=window_length,
        polyorder=polyorder,
        axis=2,
        mode='nearest'
    ).astype(np.float32)


def snv_normalization(cube, mask):
    """Standard Normal Variate normalization"""
    cube_snv = cube.copy()
    spectra = cube[mask]

    mean = np.mean(spectra, axis=1, keepdims=True)
    std = np.std(spectra, axis=1, keepdims=True) + 1e-8
    spectra_snv = (spectra - mean) / std

    cube_snv[mask] = spectra_snv
    return cube_snv.astype(np.float32)


def preprocess_cube(mat_path, return_mask=True):
    """Complete preprocessing pipeline"""
    cube = load_hsi_cube(mat_path)
    cube = minmax_normalize(cube)
    mask = get_roi_mask(cube)
    cube = sg_smoothing(cube)
    cube = snv_normalization(cube, mask)

    if return_mask:
        return cube, mask
    return cube


# ============================================================
# 5. ENHANCED PCA
# ============================================================

def sample_roi_pixels_for_pca(files, max_pixels_per_image=None):
    """Stratified pixel sampling for PCA fitting"""
    if max_pixels_per_image is None:
        max_pixels_per_image = cfg.MAX_PIXELS_PER_IMAGE_PCA

    all_pixels = []

    for i, path in enumerate(files):
        if i % 20 == 0:
            print(f"PCA sampling image {i}/{len(files)}")

        try:
            cube, mask = preprocess_cube(path)
            pixels = cube[mask]

            if len(pixels) > max_pixels_per_image:
                idx = np.random.choice(len(pixels), max_pixels_per_image, replace=False)
                pixels = pixels[idx]

            all_pixels.append(pixels)
        except Exception as e:
            print(f"Warning: Failed to process {path}: {e}")

        del cube, mask, pixels
        gc.collect()

    if len(all_pixels) == 0:
        raise ValueError("No valid pixels collected for PCA")

    return np.vstack(all_pixels).astype(np.float32)


def fit_pca(train_files, n_components=None, variance_threshold=None):
    """Fit PCA on training data with variance threshold"""
    if n_components is None:
        n_components = cfg.PCA_COMPONENTS
    if variance_threshold is None:
        variance_threshold = cfg.PCA_VARIANCE_THRESHOLD

    print("Sampling train ROI pixels for PCA...")
    pca_pixels = sample_roi_pixels_for_pca(train_files)

    print(f"PCA training data: {pca_pixels.shape}")

    # Fit PCA
    pca = PCA(n_components=n_components, random_state=cfg.RANDOM_STATE)
    pca.fit(pca_pixels)

    # Report explained variance
    total_variance = np.sum(pca.explained_variance_ratio_)
    print(f"PCA fitted. Explained variance: {total_variance:.4f}")
    print(f"Components: {pca.n_components_}")

    # Save PCA
    pca_path = os.path.join(cfg.OUTPUT_DIR, "pca_model.pkl")
    joblib.dump(pca, pca_path)
    print(f"PCA saved to {pca_path}")

    del pca_pixels
    gc.collect()

    return pca


def apply_pca_to_cube(cube, pca):
    """Apply PCA transformation to HSI cube"""
    h, w, b = cube.shape
    pixels = cube.reshape(-1, b)
    pixels_pca = pca.transform(pixels)
    cube_pca = pixels_pca.reshape(h, w, pca.n_components_)
    return cube_pca.astype(np.float32)


# ============================================================
# 6. MULTI-SCALE PATCH EXTRACTION
# ============================================================

def extract_multi_scale_patches(cube_pca, mask, patch_sizes=None, patches_per_image=None):
    """Extract multi-scale patches from ROI"""
    if patch_sizes is None:
        patch_sizes = cfg.PATCH_SIZES
    if patches_per_image is None:
        patches_per_image = cfg.PATCHES_PER_IMAGE

    h, w, c = cube_pca.shape
    all_patches = []

    # Get ROI coordinates
    coords = np.argwhere(mask)
    if len(coords) == 0:
        coords = np.argwhere(np.ones((h, w), dtype=bool))

    # Center patch (always included)
    center = (h // 2, w // 2)

    # Random patches from ROI
    n_random = min(patches_per_image - 1, len(coords))
    selected_idx = np.random.choice(len(coords), size=n_random, replace=False)

    for scale_idx, patch_size in enumerate(patch_sizes):
        half = patch_size // 2
        patches_scale = []

        # Center patch for this scale
        x1 = max(0, center[0] - half)
        y1 = max(0, center[1] - half)
        x1 = min(x1, h - patch_size)
        y1 = min(y1, w - patch_size)

        patch = cube_pca[x1:x1+patch_size, y1:y1+patch_size, :]
        if patch.shape == (patch_size, patch_size, c):
            # Pad if needed
            if patch.shape[0] < patch_size or patch.shape[1] < patch_size:
                pad_h = patch_size - patch.shape[0]
                pad_w = patch_size - patch.shape[1]
                patch = np.pad(patch, ((0, pad_h), (0, pad_w), (0, 0)), mode='reflect')
            patches_scale.append(patch)

        # Random patches
        for idx in selected_idx[:max(1, patches_per_image // len(patch_sizes))]:
            cx, cy = coords[idx]
            x1 = max(0, cx - half)
            y1 = max(0, cy - half)
            x1 = min(x1, h - patch_size)
            y1 = min(y1, w - patch_size)

            patch = cube_pca[x1:x1+patch_size, y1:y1+patch_size, :]
            if patch.shape == (patch_size, patch_size, c):
                if patch.shape[0] < patch_size or patch.shape[1] < patch_size:
                    pad_h = patch_size - patch.shape[0]
                    pad_w = patch_size - patch.shape[1]
                    patch = np.pad(patch, ((0, pad_h), (0, pad_w), (0, 0)), mode='reflect')
                patches_scale.append(patch)

        # Store patches with scale info
        for patch in patches_scale:
            all_patches.append((patch, scale_idx))

    # Return patches (resize to common size if multi-scale)
    if cfg.USE_MULTI_SCALE and len(patch_sizes) > 1:
        # Resize all to median size
        target_size = patch_sizes[len(patch_sizes) // 2]
        resized_patches = []
        for patch, scale_idx in all_patches:
            if patch.shape[0] != target_size or patch.shape[1] != target_size:
                from skimage.transform import resize
                patch = resize(patch, (target_size, target_size, c),
                              preserve_range=True, anti_aliasing=True)
            resized_patches.append(patch.astype(np.float32))
        return resized_patches
    else:
        return [p for p, _ in all_patches]


# ============================================================
# 7. DATA AUGMENTATION
# ============================================================

def augment_patch(patch):
    """Apply random augmentations to patch"""
    if not cfg.USE_AUGMENTATION:
        return patch

    augmented = patch.copy()

    # Random flip
    if cfg.AUGMENT_FLIP and np.random.random() > 0.5:
        augmented = np.flip(augmented, axis=0)
    if cfg.AUGMENT_FLIP and np.random.random() > 0.5:
        augmented = np.flip(augmented, axis=1)

    # Random rotation
    if cfg.AUGMENT_ROTATION:
        k = np.random.randint(0, 4)
        augmented = np.rot90(augmented, k, axes=(0, 1))

    return augmented


# ============================================================
# 8. DATASET BUILDING
# ============================================================

def build_patch_dataset(files, labels, pca, dataset_name="train"):
    """Build patch dataset with multi-scale extraction and augmentation"""
    X = []
    y = []
    image_ids = []
    patch_ids = []

    for i, path in enumerate(files):
        if i % 10 == 0:
            print(f"{dataset_name}: processing image {i}/{len(files)}")

        try:
            cube, mask = preprocess_cube(path)
            cube_pca = apply_pca_to_cube(cube, pca)

            patches = extract_multi_scale_patches(
                cube_pca, mask,
                patch_sizes=cfg.PATCH_SIZES,
                patches_per_image=cfg.PATCHES_PER_IMAGE
            )

            # Add augmentations for training set
            if dataset_name == "train" and cfg.USE_AUGMENTATION:
                augmented_patches = []
                for patch in patches:
                    augmented_patches.append(patch)
                    augmented_patches.append(augment_patch(patch))
                patches = augmented_patches

            X.extend(patches)
            y.extend([labels[i]] * len(patches))
            image_ids.extend([i] * len(patches))
            patch_ids.extend(range(len(patches)))

        except Exception as e:
            print(f"Warning: Failed to process {path}: {e}")

        del cube, mask, cube_pca, patches
        gc.collect()

    if len(X) == 0:
        raise ValueError(f"No patches generated for {dataset_name}")

    X = np.array(X, dtype=np.float32)
    y = np.array(y)

    # Add channel dimension for 3D CNN
    X = X[..., np.newaxis]

    # One-hot encode labels
    num_classes = len(set(labels))
    y_cat = to_categorical(y, num_classes=num_classes)

    print(f"{dataset_name} dataset: {X.shape} patches, {len(set(y))} classes")

    return X, y_cat, y, np.array(image_ids), np.array(patch_ids)


# ============================================================
# 9. LIGHTWEIGHT 3D CNN WITH SE ATTENTION
# ============================================================

def se_block_3d(x, ratio=None, block_name=""):
    """Squeeze-and-Excitation block for 3D tensors"""
    if ratio is None:
        ratio = cfg.SE_RATIO

    channels = x.shape[-1]
    se = layers.GlobalAveragePooling3D(name=f'{block_name}_se_gap')(x)
    se = layers.Dense(
        max(channels // ratio, 1),
        activation="relu",
        kernel_initializer=HeNormal(seed=cfg.RANDOM_STATE),
        name=f'{block_name}_se_fc1'
    )(se)
    se = layers.Dense(
        channels,
        activation="sigmoid",
        kernel_initializer=HeNormal(seed=cfg.RANDOM_STATE),
        name=f'{block_name}_se_fc2'
    )(se)
    se = layers.Reshape((1, 1, 1, channels), name=f'{block_name}_se_reshape')(se)

    return layers.Multiply(name=f'{block_name}_se_mul')([x, se])


def lightweight_conv_block(
    x,
    filters,
    kernel_size=(3,3,3),
    use_se=True,
    block_name=""
):

    x = layers.Conv3D(
        filters=filters,
        kernel_size=kernel_size,
        padding="same",
        use_bias=False,
        kernel_initializer=HeNormal(seed=cfg.RANDOM_STATE),
        kernel_regularizer=regularizers.l2(cfg.L2_REG),
        name=f"{block_name}_conv"
    )(x)

    x = layers.BatchNormalization(
        name=f"{block_name}_bn"
    )(x)

    x = layers.Activation(
        "relu",
        name=f"{block_name}_relu"
    )(x)

    if use_se:
        x = se_block_3d(x, block_name=block_name)

    return x


def build_lightweight_3dcnn(input_shape, num_classes, use_se=True):
    """Build lightweight 3D CNN with SE attention"""
    inputs = layers.Input(shape=input_shape, name='input')

    # Block 1
    x = lightweight_conv_block(inputs, cfg.CNN_FILTERS[0], use_se=use_se, block_name="block1")
    x = layers.MaxPooling3D((2, 2, 2), name='pool1')(x)
    x = layers.Dropout(cfg.DROPOUT_RATE, name='drop1')(x)

    # Block 2
    x = lightweight_conv_block(x, cfg.CNN_FILTERS[1], use_se=use_se, block_name="block2")
    x = layers.MaxPooling3D((2, 2, 2), name='pool2')(x)
    x = layers.Dropout(cfg.DROPOUT_RATE, name='drop2')(x)

    # Block 3
    x = lightweight_conv_block(x, cfg.CNN_FILTERS[2], use_se=use_se, block_name="block3")

    # Global pooling
    x = layers.GlobalAveragePooling3D(name='global_avg_pool')(x)

    # Dense layers
    x = layers.Dense(
        64,
        activation='relu',
        kernel_initializer=HeNormal(seed=cfg.RANDOM_STATE),
        kernel_regularizer=regularizers.l2(cfg.L2_REG),
        name='transfer_feature_layer'
    )(x)
    x = layers.Dropout(cfg.DROPOUT_RATE, name='dense_dropout')(x)

    # Output
    outputs = layers.Dense(
        num_classes,
        activation='softmax',
        kernel_initializer=HeNormal(seed=cfg.RANDOM_STATE),
        name='output'
    )(x)

    model = models.Model(inputs, outputs, name='Lightweight3DCNN_SE')

    return model


def build_model_for_tuning(hp, input_shape, num_classes):
    """Build model for hyperparameter tuning"""
    # Hyperparameters
    filters = hp.Choice('filters', [[12, 24, 36], [16, 32, 48], [24, 48, 64]])
    dropout = hp.Float('dropout', 0.2, 0.5, step=0.05)
    l2_reg = hp.Float('l2_reg', 0.0001, 0.005, step=0.0001)
    se_ratio = hp.Choice('se_ratio', [4, 8, 12])
    learning_rate = hp.Float('lr', 1e-4, 1e-3, sampling='log')

    # Temporarily update config
    cfg.CNN_FILTERS = filters
    cfg.DROPOUT_RATE = dropout
    cfg.L2_REG = l2_reg
    cfg.SE_RATIO = se_ratio
    cfg.LEARNING_RATE = learning_rate

    model = build_lightweight_3dcnn(input_shape, num_classes, use_se=True)

    model.compile(
        optimizer=Adam(learning_rate=learning_rate),
        loss='categorical_crossentropy',
        metrics=['accuracy', keras.metrics.Precision(name='precision'),
                 keras.metrics.Recall(name='recall')]
    )

    return model


# ============================================================
# 10. COMPREHENSIVE EVALUATION METRICS
# ============================================================

def compute_comprehensive_metrics(y_true, y_pred, y_pred_prob, class_names):
    """Compute comprehensive evaluation metrics"""
    num_classes = len(class_names)

    # Basic metrics
    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred, average='weighted')
    recall = recall_score(y_true, y_pred, average='weighted')
    f1 = f1_score(y_true, y_pred, average='weighted')

    # Per-class metrics
    per_class_precision = precision_score(y_true, y_pred, average=None)
    per_class_recall = recall_score(y_true, y_pred, average=None)
    per_class_f1 = f1_score(y_true, y_pred, average=None)

    # Advanced metrics
    mcc = matthews_corrcoef(y_true, y_pred)
    kappa = cohen_kappa_score(y_true, y_pred)

    # ROC-AUC (one-vs-rest for multi-class)
    y_true_onehot = to_categorical(y_true, num_classes=num_classes)
    try:
        roc_auc = roc_auc_score(y_true_onehot, y_pred_prob, average='weighted', multi_class='ovr')
    except:
        roc_auc = 0.0

    # Confusion matrix
    cm = confusion_matrix(y_true, y_pred)

    metrics = {
        'accuracy': accuracy,
        'precision_weighted': precision,
        'recall_weighted': recall,
        'f1_weighted': f1,
        'mcc': mcc,
        'kappa': kappa,
        'roc_auc_weighted': roc_auc,
        'per_class_precision': per_class_precision.tolist(),
        'per_class_recall': per_class_recall.tolist(),
        'per_class_f1': per_class_f1.tolist(),
        'confusion_matrix': cm.tolist()
    }

    return metrics


def print_metrics_report(metrics, class_names, level="Patch"):
    """Print comprehensive metrics report"""
    print(f"""
{'='*60}
{level}-LEVEL PERFORMANCE METRICS
{'='*60}
""")

    print(f"""
Overall Metrics:
  Accuracy:           {metrics['accuracy']:.4f}
  Precision (weighted): {metrics['precision_weighted']:.4f}
  Recall (weighted):    {metrics['recall_weighted']:.4f}
  F1-Score (weighted):  {metrics['f1_weighted']:.4f}
  MCC:                {metrics['mcc']:.4f}
  Kappa:              {metrics['kappa']:.4f}
  ROC-AUC (weighted): {metrics['roc_auc_weighted']:.4f}
""")

    print(f"""
Per-Class Metrics:
  {'Class':<15} {'Precision':<12} {'Recall':<12} {'F1-Score':<12}
  {'-'*51}
""")
    for i, name in enumerate(class_names):
        print(f"  {name:<15} {metrics['per_class_precision'][i]:<12.4f} "
              f"{metrics['per_class_recall'][i]:<12.4f} {metrics['per_class_f1'][i]:<12.4f}")

    print(f"""
Confusion Matrix:
  {np.array(metrics['confusion_matrix'])}
{'='*60}
""")


def plot_confusion_matrix(cm, class_names, title="Confusion Matrix", save_path=None):
    """Plot and save confusion matrix"""
    plt.figure(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=class_names, yticklabels=class_names)
    plt.title(title, fontsize=14)
    plt.ylabel('True Label', fontsize=12)
    plt.xlabel('Predicted Label', fontsize=12)
    plt.xticks(rotation=45)
    plt.tight_layout()

    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches='tight')
        print(f"Confusion matrix saved to {save_path}")

    plt.show()


def plot_roc_curves(y_true, y_pred_prob, class_names, save_path=None):
    """Plot ROC curves for each class"""
    num_classes = len(class_names)
    y_true_onehot = to_categorical(y_true, num_classes=num_classes)

    plt.figure(figsize=(12, 10))

    for i in range(num_classes):
        fpr, tpr, _ = roc_curve(y_true_onehot[:, i], y_pred_prob[:, i])
        roc_auc = auc(fpr, tpr)
        plt.plot(fpr, tpr, label=f'{class_names[i]} (AUC = {roc_auc:.3f})')

    plt.plot([0, 1], [0, 1], 'k--', linewidth=2)
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate', fontsize=12)
    plt.ylabel('True Positive Rate', fontsize=12)
    plt.title('ROC Curves - Multi-Class', fontsize=14)
    plt.legend(loc='lower right')
    plt.grid(alpha=0.3)

    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches='tight')
        print(f"ROC curves saved to {save_path}")

    plt.show()


# ============================================================
# 11. HYPERPARAMETER TUNING
# ============================================================

def run_hyperparameter_tuning(X_train, y_train_cat, X_val, y_val_cat, input_shape, num_classes):
    """Run Keras Tuner for hyperparameter optimization"""
    try:
        import keras_tuner
    except ImportError:
        print("keras-tuner not installed. Installing...")
        os.system('pip install keras-tuner')
        import keras_tuner

    tuner = keras_tuner.RandomSearch(
        hypermodel=lambda hp: build_model_for_tuning(hp, input_shape, num_classes),
        objective='val_accuracy',
        max_trials=cfg.TUNER_TRIALS,
        executions_per_trial=1,
        overwrite=True,
        directory=os.path.join(cfg.OUTPUT_DIR, 'tuner_dir'),
        project_name='hsi_tuning'
    )

    print("Starting hyperparameter tuning...")

    stop_early = EarlyStopping(
        monitor='val_loss',
        patience=5,
        restore_best_weights=True
    )

    tuner.search(
        X_train, y_train_cat,
        validation_data=(X_val, y_val_cat),
        epochs=cfg.TUNER_EPOCHS,
        callbacks=[stop_early],
        verbose=1
    )

    best_hps = tuner.get_best_hyperparameters(num_trials=1)[0]

    print(f"""
Best Hyperparameters:
  Filters: {best_hps.get('filters')}
  Dropout: {best_hps.get('dropout')}
  L2 Reg: {best_hps.get('l2_reg')}
  SE Ratio: {best_hps.get('se_ratio')}
  Learning Rate: {best_hps.get('lr')}
""")

    # Save best config
    best_config = {
        'filters': best_hps.get('filters'),
        'dropout': best_hps.get('dropout'),
        'l2_reg': best_hps.get('l2_reg'),
        'se_ratio': best_hps.get('se_ratio'),
        'learning_rate': best_hps.get('lr')
    }

    config_path = os.path.join(cfg.OUTPUT_DIR, 'best_hyperparameters.json')
    with open(config_path, 'w') as f:
        json.dump(best_config, f, indent=2)

    print(f"Best hyperparameters saved to {config_path}")

    return best_hps


# ============================================================
# 12. MAIN PIPELINE
# ============================================================

def main():
    """Main execution pipeline"""
    start_time = datetime.now()
    print(f"Pipeline started at {start_time.strftime('%Y-%m-%d %H:%M:%S')}")

    # Step 1: Extract dataset
    print(f"""
{'='*60}
STEP 1: DATASET EXTRACTION
{'='*60}
""")
    extract_dataset(cfg.ZIP_PATH, cfg.EXTRACT_PATH)
    dataset_path = find_dataset_root(cfg.EXTRACT_PATH, cfg.CLASS_NAMES)
    print(f"Dataset path: {dataset_path}")

    # Step 2: Collect file paths
    print(f"""
{'='*60}
STEP 2: COLLECTING FILE PATHS
{'='*60}
""")
    file_paths, labels = collect_file_paths(dataset_path, cfg.CLASS_NAMES)
    print(f"Total images: {len(file_paths)}")

    for cls in cfg.CLASS_NAMES:
        count = np.sum(labels == cls)
        print(f"  {cls}: {count}")

    # Encode labels
    encoder = LabelEncoder()
    labels_encoded = encoder.fit_transform(labels)

    # Step 3: Train-test split (image-level)
    print(f"""
{'='*60}
STEP 3: TRAIN-TEST SPLIT
{'='*60}
""")
    train_files, test_files, train_labels, test_labels = train_test_split(
        file_paths, labels_encoded,
        test_size=0.2, random_state=cfg.RANDOM_STATE,
        stratify=labels_encoded
    )
    print(f"Train images: {len(train_files)}")
    print(f"Test images: {len(test_files)}")

    # Step 4: Fit PCA
    print(f"""
{'='*60}
STEP 4: FITTING PCA
{'='*60}
""")
    pca = fit_pca(train_files)

    # Step 5: Build patch datasets
    print(f"""
{'='*60}
STEP 5: BUILDING PATCH DATASETS
{'='*60}
""")

    X_train, y_train_cat, y_train_raw, train_image_ids, train_patch_ids = build_patch_dataset(
        train_files, train_labels, pca, dataset_name="train"
    )

    X_test, y_test_cat, y_test_raw, test_image_ids, test_patch_ids = build_patch_dataset(
        test_files, test_labels, pca, dataset_name="test"
    )

    print(f"""
Final datasets:
  X_train: {X_train.shape}
  X_test: {X_test.shape}
""")

    # Step 6: Hyperparameter tuning (optional)
    if cfg.USE_TUNER:
        print(f"""
{'='*60}
STEP 6: HYPERPARAMETER TUNING
{'='*60}
""")

        # Split train for validation
        X_train_sub, X_val, y_train_sub, y_val = train_test_split(
            X_train, y_train_cat, test_size=0.2, random_state=cfg.RANDOM_STATE,
            stratify=np.argmax(y_train_cat, axis=1)
        )

        best_hps = run_hyperparameter_tuning(
            X_train_sub, y_train_sub, X_val, y_val,
            X_train.shape[1:], len(cfg.CLASS_NAMES)
        )

        # Update config with best params
        cfg.CNN_FILTERS = best_hps.get('filters')
        cfg.DROPOUT_RATE = best_hps.get('dropout')
        cfg.L2_REG = best_hps.get('l2_reg')
        cfg.SE_RATIO = best_hps.get('se_ratio')
        cfg.LEARNING_RATE = best_hps.get('lr')

    # Step 7: Build model
    print(f"""
{'='*60}
STEP 7: BUILDING MODEL
{'='*60}
""")

    model = build_lightweight_3dcnn(
        input_shape=X_train.shape[1:],
        num_classes=len(cfg.CLASS_NAMES),
        use_se=True
    )

    model.compile(
        optimizer=Adam(
            learning_rate=cfg.LEARNING_RATE,
            beta_1=0.9,
            beta_2=0.999
        ),
        loss='categorical_crossentropy',
        metrics=[
            'accuracy',
            keras.metrics.Precision(name='precision'),
            keras.metrics.Recall(name='recall'),
            keras.metrics.CategoricalAccuracy(name='cat_acc')
        ]
    )

    model.summary()

    # Step 8: Training callbacks
    print(f"""
{'='*60}
STEP 8: TRAINING MODEL
{'='*60}
""")

    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

    callbacks = [
        EarlyStopping(
            monitor='val_loss',
            patience=cfg.EARLY_STOP_PATIENCE,
            restore_best_weights=True,
            verbose=1
        ),
        ModelCheckpoint(
            os.path.join(cfg.OUTPUT_DIR, f'best_model_{timestamp}.keras'),
            monitor='val_accuracy',
            save_best_only=True,
            verbose=1,
            save_weights_only=False
        ),
        ReduceLROnPlateau(
            monitor='val_loss',
            factor=0.5,
            patience=7,
            min_lr=1e-7,
            verbose=1
        ),
        CSVLogger(
            os.path.join(cfg.OUTPUT_DIR, f'training_history_{timestamp}.csv'),
            separator=',',
            append=True
        )
    ]

    # Train model
    history = model.fit(
        X_train, y_train_cat,
        validation_split=cfg.VAL_SPLIT,
        epochs=cfg.EPOCHS,
        batch_size=cfg.BATCH_SIZE,
        callbacks=callbacks,
        verbose=1
    )

    # Step 9: Patch-level evaluation
    print(f"""
{'='*60}
STEP 9: PATCH-LEVEL EVALUATION
{'='*60}
""")

    y_pred_prob = model.predict(X_test, verbose=0)
    y_pred_patch = np.argmax(y_pred_prob, axis=1)
    y_true_patch = np.argmax(y_test_cat, axis=1)

    patch_metrics = compute_comprehensive_metrics(
        y_true_patch, y_pred_patch, y_pred_prob, encoder.classes_
    )
    print_metrics_report(patch_metrics, encoder.classes_, level="Patch")

    # Step 10: Image-level evaluation (majority voting)
    print(f"""
{'='*60}
STEP 10: IMAGE-LEVEL EVALUATION
{'='*60}
""")

    image_true = []
    image_pred = []
    image_pred_prob = []

    for img_id in np.unique(test_image_ids):
        idx = np.where(test_image_ids == img_id)[0]
        true_label = y_true_patch[idx][0]
        votes = y_pred_patch[idx]
        pred_label = np.bincount(votes).argmax()

        # Average probabilities for image-level
        avg_prob = np.mean(y_pred_prob[idx], axis=0)

        image_true.append(true_label)
        image_pred.append(pred_label)
        image_pred_prob.append(avg_prob)

    image_true = np.array(image_true)
    image_pred = np.array(image_pred)
    image_pred_prob = np.array(image_pred_prob)

    image_metrics = compute_comprehensive_metrics(
        image_true, image_pred, image_pred_prob, encoder.classes_
    )
    print_metrics_report(image_metrics, encoder.classes_, level="Image")

    # Step 11: Save models and artifacts
    print(f"""
{'='*60}
STEP 11: SAVING MODELS AND ARTIFACTS
{'='*60}
""")

    # Save final model
    final_model_path = os.path.join(cfg.OUTPUT_DIR, f'final_model_{timestamp}.keras')
    model.save(final_model_path)
    print(f"Model saved: {final_model_path}")

    # Save encoder
    encoder_path = os.path.join(cfg.OUTPUT_DIR, 'label_encoder.pkl')
    joblib.dump(encoder, encoder_path)
    print(f"Encoder saved: {encoder_path}")

    # Save feature extractor
    feature_extractor = models.Model(
        inputs=model.input,
        outputs=model.get_layer('transfer_feature_layer').output
    )
    feature_path = os.path.join(cfg.OUTPUT_DIR, f'feature_extractor_{timestamp}.keras')
    feature_extractor.save(feature_path)
    print(f"Feature extractor saved: {feature_path}")

    # Save training history
    history_dict = {k: [float(vi) for vi in v] for k, v in history.history.items()}
    history_path = os.path.join(cfg.OUTPUT_DIR, f'training_history_{timestamp}.json')
    with open(history_path, 'w') as f:
        json.dump(history_dict, f, indent=2)
    print(f"Training history saved: {history_path}")

    # Save all metrics
    all_metrics = {
        'patch_level': patch_metrics,
        'image_level': image_metrics,
        'config': {
            'patch_sizes': cfg.PATCH_SIZES,
            'patches_per_image': cfg.PATCHES_PER_IMAGE,
            'pca_components': cfg.PCA_COMPONENTS,
            'cnn_filters': cfg.CNN_FILTERS,
            'dropout_rate': cfg.DROPOUT_RATE,
            'batch_size': cfg.BATCH_SIZE,
            'learning_rate': cfg.LEARNING_RATE,
            'epochs': cfg.EPOCHS
        }
    }
    metrics_path = os.path.join(cfg.OUTPUT_DIR, f'evaluation_metrics_{timestamp}.json')
    with open(metrics_path, 'w') as f:
        json.dump(all_metrics, f, indent=2)
    print(f"Metrics saved: {metrics_path}")

    # Step 12: Plot results
    print(f"""
{'='*60}
STEP 12: PLOTTING RESULTS
{'='*60}
""")

    # Accuracy and loss curves
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

    ax1.plot(history.history['accuracy'], label='Train Acc', linewidth=2)
    ax1.plot(history.history['val_accuracy'], label='Val Acc', linewidth=2)
    ax1.set_xlabel('Epoch', fontsize=12)
    ax1.set_ylabel('Accuracy', fontsize=12)
    ax1.set_title('Accuracy Curves', fontsize=14)
    ax1.legend()
    ax1.grid(alpha=0.3)

    ax2.plot(history.history['loss'], label='Train Loss', linewidth=2)
    ax2.plot(history.history['val_loss'], label='Val Loss', linewidth=2)
    ax2.set_xlabel('Epoch', fontsize=12)
    ax2.set_ylabel('Loss', fontsize=12)
    ax2.set_title('Loss Curves', fontsize=14)
    ax2.legend()
    ax2.grid(alpha=0.3)

    plt.tight_layout()
    plot_path = os.path.join(cfg.OUTPUT_DIR, f'training_curves_{timestamp}.png')
    plt.savefig(plot_path, dpi=300, bbox_inches='tight')
    print(f"Training curves saved: {plot_path}")
    plt.show()

    # Confusion matrices
    plot_confusion_matrix(
        patch_metrics['confusion_matrix'],
        encoder.classes_,
        title=f'Patch-Level Confusion Matrix (Acc: {patch_metrics["accuracy"]:.3f})',
        save_path=os.path.join(cfg.OUTPUT_DIR, f'cm_patch_{timestamp}.png')
    )

    plot_confusion_matrix(
        image_metrics['confusion_matrix'],
        encoder.classes_,
        title=f'Image-Level Confusion Matrix (Acc: {image_metrics["accuracy"]:.3f})',
        save_path=os.path.join(cfg.OUTPUT_DIR, f'cm_image_{timestamp}.png')
    )

    # ROC curves
    plot_roc_curves(
        y_true_patch, y_pred_prob, encoder.classes_,
        save_path=os.path.join(cfg.OUTPUT_DIR, f'roc_curves_{timestamp}.png')
    )

    # Per-class metrics bar chart
    fig, ax = plt.subplots(figsize=(12, 6))
    x = np.arange(len(encoder.classes_))
    width = 0.25

    ax.bar(x - width, image_metrics['per_class_precision'], width, label='Precision')
    ax.bar(x, image_metrics['per_class_recall'], width, label='Recall')
    ax.bar(x + width, image_metrics['per_class_f1'], width, label='F1-Score')

    ax.set_xlabel('Class', fontsize=12)
    ax.set_ylabel('Score', fontsize=12)
    ax.set_title('Per-Class Performance Metrics (Image-Level)', fontsize=14)
    ax.set_xticks(x)
    ax.set_xticklabels(encoder.classes_, rotation=45)
    ax.legend()
    ax.grid(alpha=0.3, axis='y')

    plt.tight_layout()
    metrics_plot_path = os.path.join(cfg.OUTPUT_DIR, f'per_class_metrics_{timestamp}.png')
    plt.savefig(metrics_plot_path, dpi=300, bbox_inches='tight')
    print(f"Per-class metrics saved: {metrics_plot_path}")
    plt.show()

    # Final summary
    end_time = datetime.now()
    duration = end_time - start_time

    print(f"""
{'='*60}
PIPELINE COMPLETED SUCCESSFULLY
{'='*60}
""")
    print(f"Start time: {start_time.strftime('%Y-%m-%d %H:%M:%S')}")
    print(f"End time:   {end_time.strftime('%Y-%m-%d %H:%M:%S')}")
    print(f"Duration:   {duration}")
    print(f"""
Output directory: {cfg.OUTPUT_DIR}

Key Results:
  Patch-level Accuracy:  {patch_metrics['accuracy']:.4f}
  Image-level Accuracy:  {image_metrics['accuracy']:.4f}
  Image-level F1-Score:  {image_metrics['f1_weighted']:.4f}
  Image-level MCC:       {image_metrics['mcc']:.4f}
  Image-level Kappa:     {image_metrics['kappa']:.4f}
{'='*60}
""")

    return model, history, all_metrics


# Run the pipeline
if __name__ == "__main__":
    model, history, metrics = main()
```
